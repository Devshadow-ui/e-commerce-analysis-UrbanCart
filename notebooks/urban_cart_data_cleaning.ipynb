```python
def load_csv(folder_path):
    datasets = {}

    for file in Path(folder_path).glob("*.csv"):

        df = pd.read_csv(file)

        df.columns = (
            df.columns
            .str.strip()
            .str.lower()
            .str.replace(" ", "_")
        )

        print(f"\n{file.name}")
        print(df.isnull().sum())

        datasets[file.stem] = df

    return datasets
```

```python
def parse_messy_date(series):
    formats = ["%Y-%m-%d", "%d-%m-%Y", "%d/%m/%Y", "%m/%d/%Y", "%d %b %Y", "%Y/%m/%d"]
    result = pd.Series(pd.NaT, index=series.index)
    remaining = series.astype(str).str.strip()
    for fmt in formats:
        parsed = pd.to_datetime(remaining, format=fmt, errors="coerce")
        result = result.fillna(parsed)
        remaining = remaining.where(parsed.isna())
    return result
shipments["ship_date"] = parse_messy_date(shipments["ship_date"])
shipments["delivery_date"] = parse_messy_date(shipments["delivery_date"])
    
```

```python
def check_categories(df):
    cat_cols = df.select_dtypes(include="object").columns

    for col in cat_cols:
        print(f"{'='*50}")
        print(f"column: {col}")
        print(df[col].value_counts(dropna=False))

        #standardize
        df[col] = df[col].str.strip().str.title()

        print("\_n After:")
        print(df[col].value_counts(dropna=False))

    return df
```

```python
def load_to_db(file_path, table_name, engine):
    try:
        df = pd.read_csv(file_path, parse_dates=True)
        df.to_sql(name=table_name, con=engine, if_exists="append", index=False)
        print(f"{table_name} loaded: {len(df)} rows")
    except Exception as e:
        print(f"{table_name}: ERROR - {e}")
```
